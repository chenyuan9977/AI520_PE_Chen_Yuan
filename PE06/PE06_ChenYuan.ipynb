```python
######################################################
######### Part 1: Setup and Initial Dataset  #########
######################################################
import random
import nltk
from nltk.tokenize import word_tokenize
from textblob import TextBlob

# Download NLTK resources if not already present
try:
    nltk.data.find('tokenizers/punkt')
except nltk.downloader.DownloadError:
    nltk.download('punkt')
```

```python
######################################################
############     Part 2: Core Dataset    #############
######################################################
dataset = {
    "hello": ["Hello!", "Hi there!", "Hey!"],
    "how are you": ["I'm doing well, thanks for asking.", "I'm great, how about you?"],
    "what is your purpose": ["My purpose is to assist you with your queries.", "I am here to chat with you and help you out."],
    "what is your name": ["I don't have a name.", "You can call me Chatbot."],
    "bye": ["Goodbye!", "See you later!", "Have a great day!"]
}
```

```python
######################################################
#############   Part 3: Core Functions   #############
######################################################
def handle_user_input(user_input):
    user_input = user_input.lower()
    for pattern, responses in dataset.items():
        if pattern in user_input:
            return random.choice(responses)
    return None

def handle_ambiguous_input(user_input):
    user_tokens = set(word_tokenize(user_input.lower()))
    
    # Check for partial matches using set intersection
    for pattern, responses in dataset.items():
        pattern_tokens = set(word_tokenize(pattern))
        if user_tokens.intersection(pattern_tokens):
            return random.choice(responses)

    return "I'm not sure how to respond to that. Can you rephrase?"
```

```python
######################################################
###### Part 5: Sentiment Analysis Integration  #######
######################################################
def analyse_sentiment(text):
    blob = TextBlob(text)
    # The sentiment property returns a named tuple of the form Sentiment(polarity, subjectivity)
    # Polarity is a float within the range [-1.0, 1.0].
    # Subjectivity is a float within the range [0.0, 1.0] where 0.0 is very objective and 1.0 is very subjective.
    return blob.sentiment.polarity

def enhanced_handle_user_input(user_input):
    base_response = handle_user_input(user_input)
    
    if base_response is None:
        base_response = handle_ambiguous_input(user_input)
        
    sentiment_score = analyse_sentiment(user_input)
    
    if sentiment_score > 0.5:
        return f"{base_response} I'm glad you're feeling positive!"
    elif sentiment_score < -0.5:
        return f"{base_response} I'm sorry to hear that. I hope things get better."
    else:
        return base_response
```

```python
######################################################
#########   Part 6: Testing and Validation    ########
######################################################
def start_chat():
    print("Chatbot: Hello! I'm here to help. Type 'exit' or 'bye' to end our chat.")
    while True:
        user_input = input("You: ")
        
        if user_input.lower() in ['exit', 'bye']:
            print("Chatbot: Goodbye! Have a great day.")
            break
            
        # The main loop now uses the enhanced function
        response = enhanced_handle_user_input(user_input)
        print(f"Chatbot: {response}")

# Main execution block
if __name__ == "__main__":
    start_chat()
```
