```python
import nltk
import numpy as np
from nltk.corpus import movie_reviews
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

nltk.download('movie_reviews')

# Load the dataset
documents = [(list(movie_reviews.words(fileid)), category)
             for category in movie_reviews.categories()
             for fileid in movie_reviews.fileids(category)]
np.random.seed(42)
np.random.shuffle(documents)

texts = [" ".join(words) for words, label in documents]
labels = [1 if label == "pos" else 0 for _, label in documents]

# Split data
X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.2, random_state=42)
```

```python
# Task 1: CountVectorizer + Naive Bayes
count_vect = CountVectorizer(max_features=3000)
X_train_counts = count_vect.fit_transform(X_train)
X_test_counts = count_vect.transform(X_test)

nb = MultinomialNB()
nb.fit(X_train_counts, y_train)
nb_pred = nb.predict(X_test_counts)
print("Naive Bayes with CountVectorizer Accuracy:", accuracy_score(y_test, nb_pred))
```

```python
# Task 3: TfidfVectorizer + Naive Bayes
tfidf_vect = TfidfVectorizer(max_features=3000)
X_train_tfidf = tfidf_vect.fit_transform(X_train)
X_test_tfidf = tfidf_vect.transform(X_test)

nb_tfidf = MultinomialNB()
nb_tfidf.fit(X_train_tfidf, y_train)
nb_tfidf_pred = nb_tfidf.predict(X_test_tfidf)
print("Naive Bayes with TfidfVectorizer Accuracy:", accuracy_score(y_test, nb_tfidf_pred))
```

```python
# Task 4: SVM Classifier
svm = LinearSVC()
svm.fit(X_train_counts, y_train)
svm_pred = svm.predict(X_test_counts)
print("SVM with CountVectorizer Accuracy:", accuracy_score(y_test, svm_pred))
```
