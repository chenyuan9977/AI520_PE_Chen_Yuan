```python
########################################################
########### Task 1: Basic Text Representation ##########
########################################################
import nltk
from nltk.corpus import reuters
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Download Reuters corpus
nltk.download('reuters')

# Load Reuters corpus
documents = reuters.fileids()
corpus = [reuters.raw(doc_id) for doc_id in documents]

print(f"Number of documents: {len(corpus)}")
print("Sample document:\n", corpus[0][:500])
```

```python
########################################################
###### Task 2: Create Bag-of-Words Representation ######
########################################################
count_vectorizer = CountVectorizer()
X_bow = count_vectorizer.fit_transform(corpus)

print(f"Bag-of-Words matrix shape: {X_bow.shape}")
vocab = count_vectorizer.vocabulary_
sorted_vocab = sorted(vocab.items(), key=lambda x: x[1])
print("First 10 terms in vocabulary:", [term for term, idx in sorted_vocab[:10]])
```

```python
########################################################
######### Task 3: Create TF-IDF Representation #########
########################################################

tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(corpus)

print(f"TF-IDF matrix shape: {X_tfidf.shape}")
doc_index = 0  # Example document index
print("Non-zero TF-IDF features in document 0:", X_tfidf[doc_index].nonzero()[1])
```

```python
########################################################
############# Task 4: Stop Words Analysis ##############
########################################################
tfidf_no_stop = TfidfVectorizer(stop_words='english')
X_tfidf_no_stop = tfidf_no_stop.fit_transform(corpus)

print(f"Vocabulary size before stop words removal: {len(tfidf_vectorizer.vocabulary_)}")
print(f"Vocabulary size after stop words removal: {len(tfidf_no_stop.vocabulary_)}")
print("Number of features in document 0 before:", X_tfidf[doc_index].nnz)
print("Number of features in document 0 after:", X_tfidf_no_stop[doc_index].nnz)
```

```python
########################################################
########### Task 5: New Document Processing ############
########################################################
def process_new_document(document, vectorizer):
    """
    Process a new document using trained vectorizer.

    Args:
        document (str): Text to process
        vectorizer: Trained CountVectorizer or TfidfVectorizer

    Returns:
        sparse matrix: Document vector
    """
    return vectorizer.transform([document])

# Test with a sample news article
sample_text = "The stock market showed significant growth due to increased investor confidence."
bow_vector = process_new_document(sample_text, count_vectorizer)
tfidf_vector = process_new_document(sample_text, tfidf_vectorizer)

print("Bag-of-Words vector (non-zero features):", bow_vector.nonzero()[1])
print("TF-IDF vector (non-zero features):", tfidf_vector.nonzero()[1])
```
